```python
import numpy as np
import pandas as pd
marks=np.array([88,72,95,60,79])
print("Array:",marks)
print("Mean:",marks.mean())
print("Max:",marks.max())
print("Min:",marks.mean())
data={
    "name":["asha","Ravi","Meena","kiran","Deepa"],
    "Department":["CSE","ECE","CSE","ME","ECE"],
    "Marks":[88,72,95,60,79]
}
df=pd.DataFrame(data)
df.to_csv("Student.csv",index=False)
df.to_json("student.json",orient="records")
```
