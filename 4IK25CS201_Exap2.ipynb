```python
import pandas as pd
import numpy as np
data={
    'Name':["Amit","Bebo","Chetan","Amit","Divya","ravi","Sneha","karan","Pooja","Manish","Anjali","Suresh",
            "Kavya","Rahul","Meera","Nisha","Arjun","Divya","Rohit","Swati","Naveen","Priya","Karan","Deepa"],
    "Age":[25,np.nan,30,25,28,22,np.nan,27,26,23,29,np.nan,25,22,28,24,np.nan,28,26,23,30,27,np.nan,25],
    "Marks":[85,90,np.nan,85,78,95,8,72,np.nan,91,67,84,79,np.nan,93,88,76,82,78,np.nan,90,85,71,72],
    "City":["Mysore","Mysore","Bengaluru","Mysore",np.nan,"Dehli","Mysore","Mangalore","Bengaluru",np.nan,
            "Mysuru","Hubli","Mangalore","Dehli","Bengaluru",np.nan,"Mysore","Hubli",np.nan,"Dehli",
            "Mangalore","Bengaluru","Mangalore","Hubli"]
    }
df=pd.DataFrame(data)
print("Dataset")
print(df)
```
